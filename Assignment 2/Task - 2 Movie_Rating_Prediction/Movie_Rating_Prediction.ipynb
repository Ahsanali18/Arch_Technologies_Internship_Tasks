**Movie Rating Prediction**

Using a movie ratings dataset, build a model to predict how a user might rate a movie they haven’t seen yet. Preprocess the data, use collaborative filtering or a regression model, and evaluate its performance.

## Data Source: Kaggle
### Link: [Movielens dataset](https://www.kaggle.com/datasets/ayushimishra2809/movielens-dataset)

## Import Required Libraries

```python
from surprise import Dataset, Reader
from surprise import SVD
from surprise.model_selection import train_test_split
from surprise import accuracy
import numpy as np
import pandas as pd
```

## Load Dataset

```python
movie_ratings_data = pd.read_csv("ratings.csv")

movie_ratings_data.head()
```

## Prepare Data for Model

```python
reader = Reader(rating_scale=(1,5))

surprise_data = Dataset.load_from_df(
    movie_ratings_data[['userId','movieId','rating']],
    reader
)

train_set, test_set = train_test_split(surprise_data, test_size=0.2, random_state=42)
```

## Train SVD Model
## SVD Rating Prediction Formula

The predicted rating is calculated as:

$$
\hat{r}_{ui} = \mu + b_u + b_i + q_i^T p_u
$$

Where:
- $\mu$ = global average rating  
- $b_u$ = user bias  
- $b_i$ = movie bias  
- $q_i^T p_u$ = dot product of latent feature vectors  

```python
svd_model = SVD()
svd_model.fit(train_set)
```

## Make Predictions

```python
predictions = svd_model.test(test_set)
```

## Evaluate Model
## Model Evaluation - RMSE

The Root Mean Squared Error (RMSE) is computed as:

$$
RMSE = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_{true} - y_{pred})^2}
$$

Where:
- $y_{true}$ = actual rating  
- $y_{pred}$ = predicted rating  
- $n$ = total number of predictions  

```python
rmse_value = accuracy.rmse(predictions)
print("RMSE:", rmse_value)
```

## Predict Rating for Specific User

```python
predicted_rating = svd_model.predict(uid=1, iid=50)

print("Predicted Rating:", predicted_rating.est)
```
